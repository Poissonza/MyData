```python
from TornAPI.Torn import User
from pyspark.sql.types import StructField, StructType, IntegerType, StringType, ArrayType, BooleanType
from pyspark.sql.functions import lit
import datetime as dt

user_api = User(dbutils.secrets.get("Personal", "TornAPI"))
```

```python
ec_data = user_api.get_enlisted_cars()
```

```python
schema = StructType([
    StructField("id", IntegerType()),
    StructField("name", StringType()),
    StructField("item_id", IntegerType()),
    StructField("car_item_id", IntegerType()),
    StructField("item_name", StringType()),
    StructField("top_speed", IntegerType()),
    StructField("acceleration", IntegerType()),
    StructField("bracking", IntegerType()),
    StructField("handling", IntegerType()),
    StructField("safety", IntegerType()),
    StructField("dirt", IntegerType()),
    StructField("tarmac", IntegerType()),
    StructField("class", StringType()),
    StructField("worth", IntegerType()),
    StructField("points_spent", IntegerType()),
    StructField("races_entered", IntegerType()),
    StructField("races_won", IntegerType()),
    StructField("is_removed", BooleanType()),
    StructField("parts", ArrayType(IntegerType()))
])

sp_ec_data = spark.createDataFrame(ec_data["enlistedcars"], schema=schema)
sp_ec_data = sp_ec_data.withColumn("timeStamp", lit(dt.datetime.now()))

sp_ec_data.write.format("delta").mode("append").saveAsTable("torn.user.enlisted_cars")
```
