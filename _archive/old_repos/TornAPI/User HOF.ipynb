```python
from TornAPI.Torn import User
from pyspark.sql.functions import lit
from pyspark.sql.types import StructField, StructType, IntegerType, StringType, LongType
import datetime as dt

user_api = User(dbutils.secrets.get("Personal", "TornAPI"))
```

```python
hof_data = user_api.get_hof()

# sp_hof = spark.createDataFrame([hof_data["hof"]])

# sp_hof.display()
hof_data["hof"]
```

```python
schema = StructType([
    StructField("attacks", StructType([
        StructField("value", LongType()),
        StructField("rank", IntegerType())
    ])),
    StructField("busts", StructType([
        StructField("value", LongType()),
        StructField("rank", IntegerType())
    ])),
        StructField("defends", StructType([
        StructField("value", LongType()),
        StructField("rank", IntegerType())
    ])),
    StructField("networth", StructType([
        StructField("value", LongType()),
        StructField("rank", IntegerType())
    ])),
    StructField("offences", StructType([
        StructField("value", LongType()),
        StructField("rank", IntegerType())
    ])),
    StructField("revives", StructType([
        StructField("value", LongType()),
        StructField("rank", IntegerType())
    ])),
    StructField("level", StructType([
        StructField("value", LongType()),
        StructField("rank", IntegerType())
    ])),
    StructField("rank", StructType([
        StructField("value", LongType()),
        StructField("rank", IntegerType())
    ])),
    StructField("awards", StructType([
        StructField("value", LongType()),
        StructField("rank", IntegerType())
    ])),
    StructField("racing_skill", StructType([
        StructField("value", LongType()),
        StructField("rank", IntegerType())
    ])),
    StructField("racing_points", StructType([
        StructField("value", LongType()),
        StructField("rank", IntegerType())
    ])),
    StructField("racing_wins", StructType([
        StructField("value", LongType()),
        StructField("rank", IntegerType())
    ])),
    StructField("travel_time", StructType([
        StructField("value", LongType()),
        StructField("rank", IntegerType())
    ])),
    StructField("battle_stats", StructType([
        StructField("value", LongType()),
        StructField("rank", IntegerType())
    ])),
    StructField("working_stats", StructType([
        StructField("value", LongType()),
        StructField("rank", IntegerType())
    ]))
])

sp_hof = spark.createDataFrame([hof_data["hof"]], schema)
sp_hof = sp_hof.withColumn("timeStamp", lit(dt.datetime.now()))

sp_hof.write.format("delta").mode("append").saveAsTable("torn.user.hof")
```
