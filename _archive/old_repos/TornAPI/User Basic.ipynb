```python
from TornAPI.Torn import User
from pyspark.sql.functions import lit
import datetime as dt

user_api = User(dbutils.secrets.get("Personal", "TornAPI"))
```

```python
user_basic = user_api.get_basic()["profile"]

sp_basic = spark.createDataFrame([user_basic])

sp_basic = sp_basic.withColumn("timeStamp", lit(dt.datetime.now()))

sp_basic.write.format("delta").mode("append").saveAsTable("torn.user.basic")
```
