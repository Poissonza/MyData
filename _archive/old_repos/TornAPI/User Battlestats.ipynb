```python
from TornAPI.Torn import User
from pyspark.sql.types import StructField, StructType, StringType, IntegerType, ArrayType
from pyspark.sql.functions import lit
import datetime as dt

user_api = User(dbutils.secrets.get("Personal", "TornAPI"))
```

```python
bs_data = user_api.get_battlestats()["battlestats"]

# sp_bs_data = spark.createDataFrame([bs_data])

# sp_bs_data.display()
bs_data
```

```python
schema = StructType([
    StructField("strength", StructType([
        StructField("value", IntegerType()),
        StructField("modifier", IntegerType()),
        StructField("modifiers", ArrayType(
            StructType([
                StructField("effect", StringType()),
                StructField("value", IntegerType()),
                StructField("type", StringType())
            ])
        ))
    ])),

    StructField("defense", StructType([
        StructField("value", IntegerType()),
        StructField("modifier", IntegerType()),
        StructField("modifiers", ArrayType(
            StructType([
                StructField("effect", StringType()),
                StructField("value", IntegerType()),
                StructField("type", StringType())
            ])
        ))
    ])),

    StructField("speed", StructType([
        StructField("value", IntegerType()),
        StructField("modifier", IntegerType()),
        StructField("modifiers", ArrayType(
            StructType([
                StructField("effect", StringType()),
                StructField("value", IntegerType()),
                StructField("type", StringType())
            ])
        ))
    ])),

    StructField("dexterity", StructType([
        StructField("value", IntegerType()),
        StructField("modifier", IntegerType()),
        StructField("modifiers", ArrayType(
            StructType([
                StructField("effect", StringType()),
                StructField("value", IntegerType()),
                StructField("type", StringType())
            ])
        ))
    ])),
    StructField("total", IntegerType())
])

sp_data = spark.createDataFrame([bs_data], schema= schema)

sp_data = sp_data.withColumn("timeStamp", lit(dt.datetime.now()))

sp_data.display()
```
