```python
from TornAPI.Torn import User
from pyspark.sql.functions import lit
import datetime as dt

user_api = User(dbutils.secrets.get("Personal", "TornAPI"))
```

```python
sp_data = spark.createDataFrame([user_api.get_education()["education"]])

sp_data = sp_data.withColumn("timeStamp", lit(dt.datetime.now()))

sp_data.write.format("delta").mode("append").saveAsTable("torn.user.education")
```
