```python
from TornAPI.Torn import User
import datetime as dt
from pyspark.sql.functions import lit

user_api = User(dbutils.secrets.get("Personal", "TornAPI"))
```

```python
user_data = user_api.get_base_user()

sp_user = spark.createDataFrame([user_data["profile"]])

sp_user = sp_user.withColumn("timeStamp", lit(dt.datetime.now()))
sp_user.write.format("delta").mode("append").saveAsTable("torn.user.base")
```
