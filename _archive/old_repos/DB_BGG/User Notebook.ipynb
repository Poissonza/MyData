```python
from bggAPI.BGGApi import BGGUsers
from bggAPI.BGGDataCleaner import BGGUserCleaner



bgg = BGGUsers()

# sp_usr_df = spark.read.table("boardgame.boardgamegeek.src__bgg_users")

# df_usr_list = [data[0] for data in sp_usr_df.select("user_name").collect()]

usr_list= ["scod101", "MikeDT", "Simon75", "terriswheel", "Jules969", "VikingZ", "Deathpaw", "Inconmon"]
final_list = []
for user in usr_list:    
    usr_data = bgg.get_data(user)
    if usr_data is not None:
        cleaner = BGGUserCleaner(usr_data)
        final_list.append(cleaner.clean_user())
        
if len(final_list)> 0:
    sp_usr_df = spark.createDataFrame(final_list)
    sp_usr_df.write.format("delta").mode("append").saveAsTable("boardgame.boardgamegeek.src__bgg__users")
```
