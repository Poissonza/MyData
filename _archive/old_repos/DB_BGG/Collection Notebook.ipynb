```python
from bggAPI.BGGApi import BGGCollection
from bggAPI.BGGDataCleaner import BGGCollectionCleaner
```

```python
usr_df = spark.read.table("boardgame.boardgamegeek.src__bgg_users")

row = usr_df.select("user_name").collect()

user_list = [user["user_name"] for user in row]

fin_data = []
col_api = BGGCollection()

for user in user_list:
    data, user_id = col_api.get_collection(user)
    cleaner = BGGCollectionCleaner(data)
    fin_data = fin_data + cleaner.clean_collection(user_id)

sp_df = spark.createDataFrame(fin_data)

sp_df.write.format("delta").mode("overwrite").saveAsTable("boardgame.boardgamegeek.src__bgg__collections")

```
